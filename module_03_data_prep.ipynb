# Notebook: Module 3 - Data Preparation and Feature Engineering

```python
import polars as pl
import numpy as np

# Load example dataset
from sklearn.datasets import fetch_california_housing
raw = fetch_california_housing(as_frame=True)
df = pl.from_pandas(raw.frame)
df.head()
```

# Handle Missing Values

# (Note: no missing values in this dataset, but here's how you would handle it)

```python
df = df.fill_null(strategy='mean')
```

# Feature Scaling and Normalization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled = scaler.fit_transform(raw.frame)
scaled_df = pl.DataFrame(scaled, schema=raw.feature_names)
scaled_df.head()
```

# Feature Creation

```python
df = df.with_columns([
    (df['AveRooms'] / df['HouseAge']).alias('RoomsPerAge')
])
df.select(['AveRooms', 'HouseAge', 'RoomsPerAge']).head()
```
