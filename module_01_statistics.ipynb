# Notebook: Module 1 - Essential Mathematics and Statistics

```python
import numpy as np
import polars as pl
import altair as alt
```

# Descriptive Statistics

```python
sample = np.array([10, 20, 20, 40, 50])
print("Mean:", np.mean(sample))
print("Median:", np.median(sample))
print("Standard Deviation:", np.std(sample))
```

# Probability Basics

```python
coin_flips = np.random.choice(["Heads", "Tails"], size=10)
print("Coin flips:", coin_flips)
```

# Correlation and Covariance

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
print("Correlation:", np.corrcoef(x, y))
print("Covariance:", np.cov(x, y))
```

# Simple Linear Algebra

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 3]])
print("Matrix product:", np.dot(A, B))
```

# Altair Visualization

```python
import pandas as pd
sample_df = pd.DataFrame({"value": sample})
chart = alt.Chart(sample_df).mark_bar().encode(
    x="value:Q",
    y="count():Q"
)
chart
```
