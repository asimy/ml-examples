# Notebook: Module 5 - Underfitting, Overfitting, and Regularization

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

np.random.seed(0)
X = np.sort(np.random.rand(40, 1) * 2 - 1, axis=0)
y = X**3 - 0.5 * X + np.random.normal(0, 0.1, X.shape)
```

# Polynomial Regression (overfitting vs underfitting)

```python
poly = PolynomialFeatures(degree=15)
X_poly = poly.fit_transform(X)
model = LinearRegression().fit(X_poly, y)
y_pred = model.predict(X_poly)
```

# Visualize

```python
plt.scatter(X, y, color='blue')
plt.plot(X, y_pred, color='red')
plt.title('Polynomial Regression (Degree 15)')
plt.show()
```

# Ridge Regularization

```python
ridge = Ridge(alpha=1.0).fit(X_poly, y)
y_ridge = ridge.predict(X_poly)
plt.scatter(X, y)
plt.plot(X, y_ridge, color='green')
plt.title('With Ridge Regularization')
plt.show()
```
